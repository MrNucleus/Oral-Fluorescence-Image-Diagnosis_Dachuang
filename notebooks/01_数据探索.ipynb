# 数据探索与分析

## 目的
1. 探索数据分布
2. 可视化样本
3. 分析图像质量

```python
import sys; sys.path.append("..")
import os, cv2, numpy as np, matplotlib.pyplot as plt
from src.config import Config
cfg = Config()
print(f"数据目录: {cfg.data.data_root}")
```

```python
# 统计各类别数量
if not os.path.exists(cfg.data.data_root):
    print("请将数据放入 data/raw/")
else:
    for c in cfg.data.class_names:
        p = os.path.join(cfg.data.data_root, c)
        n = len(os.listdir(p)) if os.path.exists(p) else 0
        print(f"  {c}: {n}")
```

```python
# 加载数据集
from src.data.dataset import OralFluorescenceDataset
from src.data.transforms import get_val_transforms
sp = os.path.join(cfg.data.split_dir, cfg.data.train_split)
if os.path.exists(sp):
    ds = OralFluorescenceDataset(cfg.data.data_root, sp, transform=get_val_transforms(), class_names=cfg.data.class_names)
    print(f"Train samples: {len(ds)}")
```

```python
# 可视化样本
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i, ax in enumerate(axes.flat):
    if i >= min(len(ds), 10): break
    img, lbl = ds[i]
    img = img.permute(1,2,0).numpy()
    img = np.clip(img * np.array([0.229,0.224,0.225]) + np.array([0.485,0.456,0.406]), 0, 1)
    ax.imshow(img); ax.set_title(cfg.data.class_names[lbl]); ax.axis("off")
plt.tight_layout(); plt.show()
```
